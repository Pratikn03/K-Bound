# 08 · ELARA / RGA — the Worked Multimodal Instantiation

K-Bound generalises **ELARA**: reliability-gated multimodal anomaly fusion. The clearest instance
of the knowability boundary is MVTec-3D: gating does **no harm on clean** data and **recovers**
large accuracy **under modality failure** — exactly the knowably-helpful regime.


```python
import os, json, glob, numpy as np
REPO=os.path.abspath(os.getcwd())
while REPO!='/' and not os.path.exists(os.path.join(REPO,'scripts','rebuild_kbound.sh')): REPO=os.path.dirname(REPO)
RES=os.path.join(REPO,'experiments','kbound','results'); FIG=os.path.join(REPO,'docs','research','kbound','figures')
ARCH=os.path.join(REPO,'experiments','elara_u','score_archive'); ELA=os.path.join(REPO,'experiments','elara_u')
from IPython.display import Image, display
def jload(p):
    fp=p if os.path.isabs(p) else os.path.join(RES,p)
    return json.load(open(fp)) if os.path.exists(fp) else {}
def fig(fn,w=720):
    p=os.path.join(FIG,fn)
    if os.path.exists(p): display(Image(filename=p,width=w))
print('repo:',REPO)
```

```python
m=jload(os.path.join(ELA,'multimodal_reliability_results_mvtec3d.json'))
if m:
    print('### MVTec-3D (D23)')
    for reg,vals in m.get('regimes',{}).items(): print(' ',reg,'->',vals.get('mean_auroc'))
    print(' reliability_validated:',m.get('reliability_validated'))
s=jload(os.path.join(ELA,'shift_stress_ablation.json'))
if s:
    print('\n### Shift-stress crossover (reliability routing vs val-select)')
    for k in ['severity_0.0','severity_3.0']:
        c=s.get(k,{}).get('contrasts',{}).get('rel_vs_val_select',{}); print('  %s: mean=%.4f  CI_low>0=%s'%(k,c.get('mean',float('nan')),c.get('ci_low_gt_0')))
```
